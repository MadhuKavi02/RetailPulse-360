```python
# ═══════════════════════════════════════════════════════════════════════════════
# RetailPulse — KartZone India Pvt. Ltd.
#  PRODUCTS TABLE GENERATOR 
# ═══════════════════════════════════════════════════════════════════════════════
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random
import warnings
warnings.filterwarnings('ignore')

np.random.seed(43)
random.seed(43)

print("="*65)
print("  KARTZONE INDIA — PRODUCTS TABLE GENERATOR")
print("="*65)

# Concern 4: Enterprise-style Product ID
# Concern 6: Variant logic for unique product names

CAT_VARIANTS = {
    'Electronics':    ['Electronics','electronics','ELECTRONICS','Electronis','Electronic'],
    'Fashion':        ['Fashion','fashion','FASHION','Fasion','Clothing'],
    'Home & Kitchen': ['Home & Kitchen','home & kitchen','HOME & KITCHEN','Home','H&K'],
    'Beauty':         ['Beauty','beauty','BEAUTY','Beuty','Personal Care']
}
CAT_W = [35,30,20,15]

SUBCATS = {
    'Electronics':    ['Mobile Phone','Laptop','Tablet','Earphones','Smart Watch',
                       'Camera','Printer','WiFi Router','Power Bank','Smart TV',
                       'Keyboard','Mouse','External Hard Disk','Pen Drive','Bluetooth Speaker'],
    'Fashion':        ['Saree','Kurta','Jeans','T-Shirt','Shoes','Handbag',
                       'Jacket','Ethnic Wear','Watch','Sunglasses','Leggings',
                       'Kurti','Blazer','Sandals','Belt'],
    'Home & Kitchen': ['Cookware Set','Bedsheet','Curtains','Storage Box','LED Lamp',
                       'Pressure Cooker','Mixer Grinder','Air Purifier','Carpet','Wall Clock',
                       'Water Bottle','Lunch Box','Cutting Board','Knife Set','Towel Set'],
    'Beauty':         ['Face Cream','Shampoo','Lipstick','Perfume','Sunscreen',
                       'Hair Oil','Foundation','Moisturizer','Face Serum','Body Lotion',
                       'Eye Liner','Kajal','Face Wash','Body Scrub','Nail Polish']
}

BRANDS = {
    'Electronics':    ['Samsung','Apple','OnePlus','Xiaomi','boAt','Sony',
                       'LG','HP','Dell','Realme','Lenovo','Asus','JBL','Noise','Fire-Boltt'],
    'Fashion':        ['Fabindia','Biba','Zara','H&M','Puma','Nike',
                       'Adidas','Titan','Ray-Ban','Peter England','Arrow','Van Heusen',
                       'Aurelia','Global Desi','W'],
    'Home & Kitchen': ['Prestige','Hawkins','Nilkamal','Philips','Bajaj',
                       'TTK','Cello','Milton','Borosil','Pigeon','Wonderchef',
                       'Solimo','Amazon Basics','Kuber Industries','Story@Home'],
    'Beauty':         ['Lakme','Nykaa','Mamaearth','Himalaya','Biotique',
                       "L'Oreal","Pond's",'Dove','Garnier','WOW','Plum',
                       'mCaffeine','Minimalist','Dot & Key','Re\'equil']
}

# Concern 2: Realistic weight ranges per subcategory
WEIGHT_MAP = {
    'Mobile Phone':0.2,'Laptop':2.5,'Tablet':0.6,'Earphones':0.1,'Smart Watch':0.1,
    'Camera':0.5,'Printer':5.0,'WiFi Router':0.4,'Power Bank':0.3,'Smart TV':15.0,
    'Keyboard':0.8,'Mouse':0.1,'External Hard Disk':0.2,'Pen Drive':0.02,'Bluetooth Speaker':0.5,
    'Saree':0.4,'Kurta':0.3,'Jeans':0.5,'T-Shirt':0.2,'Shoes':0.8,'Handbag':0.5,
    'Jacket':0.7,'Ethnic Wear':0.5,'Watch':0.2,'Sunglasses':0.1,'Leggings':0.2,
    'Kurti':0.3,'Blazer':0.6,'Sandals':0.5,'Belt':0.2,
    'Cookware Set':3.0,'Bedsheet':1.5,'Curtains':1.0,'Storage Box':1.2,'LED Lamp':0.8,
    'Pressure Cooker':2.5,'Mixer Grinder':4.0,'Air Purifier':8.0,'Carpet':3.0,'Wall Clock':0.8,
    'Water Bottle':0.5,'Lunch Box':0.4,'Cutting Board':0.5,'Knife Set':1.0,'Towel Set':0.8,
    'Face Cream':0.1,'Shampoo':0.3,'Lipstick':0.05,'Perfume':0.2,'Sunscreen':0.1,
    'Hair Oil':0.2,'Foundation':0.1,'Moisturizer':0.15,'Face Serum':0.05,'Body Lotion':0.3,
    'Eye Liner':0.02,'Kajal':0.02,'Face Wash':0.2,'Body Scrub':0.3,'Nail Polish':0.05
}

COST_RANGES = {
    # Electronics
    'Mobile Phone': (8000, 45000),
    'Laptop': (30000, 75000),
    'Tablet': (10000, 40000),
    'Earphones': (500, 5000),
    'Smart Watch': (1500, 12000),
    'Camera': (15000, 60000),
    'Printer': (4000, 25000),
    'WiFi Router': (800, 5000),
    'Power Bank': (500, 2500),
    'Smart TV': (20000, 70000),
    'Keyboard': (500, 4000),
    'Mouse': (300, 3000),
    'External Hard Disk': (2500, 9000),
    'Pen Drive': (200, 1500),
    'Bluetooth Speaker': (800, 8000),

    # Fashion
    'Saree': (500, 8000),
    'Kurta': (400, 3000),
    'Jeans': (700, 3500),
    'T-Shirt': (200, 1500),
    'Shoes': (800, 6000),
    'Handbag': (700, 7000),
    'Jacket': (1000, 6000),
    'Ethnic Wear': (800, 7000),
    'Watch': (800, 8000),
    'Sunglasses': (300, 5000),
    'Leggings': (150, 1000),
    'Kurti': (400, 3000),
    'Blazer': (1200, 7000),
    'Sandals': (500, 3500),
    'Belt': (200, 1500),

    # Home & Kitchen
    'Cookware Set': (1000, 6000),
    'Bedsheet': (400, 3000),
    'Curtains': (500, 4000),
    'Storage Box': (150, 1500),
    'LED Lamp': (300, 2500),
    'Pressure Cooker': (1200, 5000),
    'Mixer Grinder': (2000, 7000),
    'Air Purifier': (5000, 20000),
    'Carpet': (800, 6000),
    'Wall Clock': (300, 2500),
    'Water Bottle': (100, 800),
    'Lunch Box': (150, 1000),
    'Cutting Board': (100, 800),
    'Knife Set': (400, 2500),
    'Towel Set': (300, 2000),

    # Beauty
    'Face Cream': (100, 1000),
    'Shampoo': (100, 800),
    'Lipstick': (150, 1200),
    'Perfume': (500, 5000),
    'Sunscreen': (150, 1200),
    'Hair Oil': (80, 600),
    'Foundation': (200, 2000),
    'Moisturizer': (150, 1200),
    'Face Serum': (300, 2500),
    'Body Lotion': (150, 1000),
    'Eye Liner': (100, 800),
    'Kajal': (80, 500),
    'Face Wash': (100, 800),
    'Body Scrub': (200, 1200),
    'Nail Polish': (80, 500)
}
MARKUP = {
    'Electronics':    (1.20,1.60),
    'Fashion':        (1.50,3.00),
    'Home & Kitchen': (1.30,2.00),
    'Beauty':         (1.40,2.50)
}

# Concern 6: Variant attributes per subcategory
VARIANTS = {
    'Mobile Phone':      [('64GB','Black'),('128GB','Blue'),('256GB','Silver'),('128GB','Gold'),('256GB','Black')],
    'Laptop':            [('8GB RAM 512GB SSD','Grey'),('16GB RAM 1TB SSD','Silver'),('8GB RAM 256GB SSD','Black')],
    'Tablet':            [('32GB WiFi','Black'),('64GB 4G','Silver'),('128GB WiFi','Grey')],
    'Earphones':         [('Wired','Black'),('Wireless','White'),('True Wireless','Blue')],
    'Smart Watch':       [('42mm','Black'),('44mm','Silver'),('40mm','Rose Gold')],
    'Shoes':             [('Size 6','Black'),('Size 7','Brown'),('Size 8','White'),('Size 9','Navy'),('Size 10','Black')],
    'T-Shirt':           [('S','White'),('M','Black'),('L','Blue'),('XL','Grey'),('XXL','Red')],
    'Jeans':             [('28W Slim','Blue'),('30W Regular','Black'),('32W Straight','Grey'),('34W Skinny','Dark Blue')],
    'Lipstick':          [('Matte','Red'),('Glossy','Pink'),('Matte','Nude'),('Satin','Berry')],
    'Foundation':        [('Shade 01','Light'),('Shade 03','Medium'),('Shade 05','Tan'),('Shade 07','Deep')],
    'Saree':             [('Silk','Blue'),('Cotton','Red'),('Georgette','Green'),('Chiffon','Pink')],
    'Kurti':             [('S','Yellow'),('M','Orange'),('L','Purple'),('XL','Green')],
    'Bedsheet':          [('Single','White'),('Double','Blue'),('King Size','Grey'),('Queen Size','Beige')],
}

# Concern 5: Product lifecycle status (more realistic than just Yes/No)
LIFECYCLE = ['Active','Active','Active','Active','Discontinued','Upcoming','Seasonal','Clearance']

MISSING = [None,np.nan,'NA','N/A','','null','-']

def messy_dt(dt):
    return dt.strftime(random.choice(['%Y-%m-%d','%d/%m/%Y','%d-%m-%Y','%m/%d/%Y']))

def messy_price(p, miss_prob=0.08):
    if random.random() < miss_prob: return random.choice(MISSING[:4])
    return random.choices(
        [p, f'₹{p}', f'Rs.{p}', f'INR {p}', str(p)],
        weights=[55,15,12,8,10]
    )[0]

def get_weight(subcat, cat):
    base = WEIGHT_MAP.get(subcat, 1.0)
    noise = random.uniform(0.85, 1.15)
    w = round(base * noise, 2)
    return random.choices([w, None, f'{w} kg', str(w)], weights=[70,10,10,10])[0]

print("\n  Generating 800 product records...")

products = []
counter = {'Electronics':1000,'Fashion':2000,'Home & Kitchen':3000,'Beauty':4000}
cat_prefix = {'Electronics':'ELEC','Fashion':'FASH','Home & Kitchen':'HOME','Beauty':'BEAU'}

for _ in range(800):
    cat_key = random.choices(list(CAT_VARIANTS.keys()), weights=CAT_W)[0]
    cat_val = random.choice(CAT_VARIANTS[cat_key])
    sub_cat = random.choice(SUBCATS[cat_key])
    brand   = random.choice(BRANDS[cat_key])

    # Concern 4: Enterprise-style Product ID
    pid_num = counter[cat_key]
    counter[cat_key] += 1
    product_id = f"{cat_prefix[cat_key]}-{pid_num}"

    # Concern 6: Add variant to product name
    variant_list = VARIANTS.get(sub_cat, None)
    if variant_list:
        v = random.choice(variant_list)
        product_name = f"{brand} {sub_cat} {' '.join(v)}"
    else:
        product_name = f"{brand} {sub_cat}"

    # Pricing
    cmin, cmax = COST_RANGES[sub_cat]
    cost = round(random.uniform(cmin, cmax), 2)
    mmin, mmax = MARKUP[cat_key]
    mrp = round(cost * random.uniform(mmin, mmax), 2)

    # Small number of clearance products
    if random.random() < 0.02:
      mrp = round(cost * random.uniform(1.01, 1.08), 2)

    # ~4% negative cost
    cost_display = cost

    # Concern 7: Add Discount_Pct and Selling_Price
    if cat_key == 'Electronics':
        disc = random.choices([0,5,10,15,20,25,30,35,40],
                               weights=[10,10,15,20,20,12,8,3,2])[0]
    elif cat_key == 'Fashion':
        disc = random.choices([0,5,10,15,20,25,30,40,50],
                               weights=[10,10,15,20,15,12,8,5,5])[0]
    else:
        disc = random.choices([0,5,10,15,20,25],
                               weights=[25,20,25,15,10,5])[0]

    selling_price = round(mrp * (1 - disc/100), 2)

    disc_fmt = random.choices([disc, float(disc), f'{disc}%', None],
                               weights=[60,10,20,10])[0]
    sp_fmt   = random.choices([selling_price, f'₹{selling_price}', None],
                               weights=[65,25,10])[0]

    # Launch date
    launch = datetime(2020,1,1) + timedelta(
        days=random.randint(0,(datetime(2024,12,31)-datetime(2020,1,1)).days)
    )
    launch_fmt = random.choices([messy_dt(launch),None,'NA',''],weights=[80,10,5,5])[0]

    # Rating — newer = lower
    days_old = (datetime.now() - launch).days
    base_r = min(5.0, 2.5 + days_old/1000)
    rating = round(np.clip(random.gauss(base_r, 0.5), 1.0, 5.0), 1)

    rating_fmt = random.choices(
    [rating, None, str(rating)],
    weights=[80, 10, 10]
)[0]

    # Stock
    base_stock = random.randint(0,500)
    stock_fmt = random.choices(
    [base_stock, float(base_stock), str(base_stock), None, 0],
    weights=[65,10,10,10,5]
)[0]

    # Weight — realistic per subcategory (Concern 2)
    w_fmt = get_weight(sub_cat, cat_key)

    # Concern 5: Lifecycle status
    lifecycle = random.choice(LIFECYCLE)
    lifecycle_fmt = random.choices(
        [lifecycle, lifecycle.lower(), lifecycle.upper(), None],
        weights=[65,12,12,11]
    )[0]

    products.append({
        'Product_ID':         product_id,
        'Product_Name':       product_name,
        'Category':           cat_val,
        'Sub_Category':       sub_cat,
        'Brand':              brand,
        'MRP':                messy_price(mrp),
        'Cost_Price':         messy_price(cost_display, miss_prob=0.08),
        'Discount_Pct':       disc_fmt,
        'Selling_Price':      sp_fmt,
        'Launch_Date':        launch_fmt,
        'Rating':             rating_fmt,
        'Stock_Units':        stock_fmt,
        'Weight_KG':          w_fmt,
        'Product_Status':     lifecycle_fmt
    })

df = pd.DataFrame(products)

# Add a small number of duplicate records for cleaning practice
dupes = df.sample(15, random_state=43)
df = pd.concat([df, dupes], ignore_index=True)
df.to_csv('KartZone_Products_V2.csv', index=False)

print(f"\n  Rows: {len(df):,} | Cols: {len(df.columns)} | Nulls: {df.isnull().sum().sum():,} | Dupes: {df.duplicated().sum()}")
print(f"  Saved → KartZone_Products_V2.csv ✓")
```
